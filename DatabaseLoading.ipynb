```python
import pandas as pd
import sqlite3

# 1. Read the CSV file into a Pandas DataFrame
csv_file_path = 'D:\\Codebase\\HCL\\ProductTable.csv'
df = pd.read_csv(csv_file_path)

# 2. Connect to the SQLite database
# (This will create the file 'my_database.db' if it doesn't exist)
db_connection = sqlite3.connect('sales.db')

# 3. Write the data to a new table in the database
table_name = 'Product'
df.to_sql(table_name, db_connection, if_exists='replace', index=False)

# 4. Close the connection
db_connection.close()

print(f"Successfully converted {csv_file_path} to {table_name} in my_database.db")
```

```python
connection = sqlite3.connect('sales.db')
query = """
SELECT 
    *
FROM 
    OrderTable O
LEFT JOIN 
    User U ON O.User_ID = U.User_ID
LEFT JOIN 
    Product P ON O.Product_ID = P.Product_ID;
"""
df = pd.read_sql_query(query, connection)
connection.close()
print(df.tail())
```
